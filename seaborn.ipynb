Data Visualization With Seaborn

Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for drawing
attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few
lines of code. In this lesson, we will cover the basics of Seaborn, including creating various types of plots and
customizing them.

```python
import seaborn as sns
```

```python
### Basic Plotting with Seaborn
tips=sns.load_dataset('tips')
tips
```

```python
## create a scatter plot
import matplotlib.pyplot as plt

sns.scatterplot(x='total_bill', y='tip',data=tips)
plt.title("Scatter Plot of Total bills vs Tips")
```

```python
## Line Plot

sns.lineplot(x='size',y='total_bill',data=tips)
plt.title("Line Plot of Total bill by size")
plt.show()
```

```python
## Categorical Plots
## Bar Plot
sns.barplot(x='day',y='total_bill',data=tips)
plt.title("Bar Plot of Total bill by Day")
plt.show()
```

```python
## Box Plot
sns.boxplot(x='day',y='total_bill',data=tips)
```

```python
## Violin Plot

sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
### Histogram
sns.histplot(tips['total_bill'], bins=10, kde=True)
```

```python
## KDE Plot

sns.kdeplot(tips['total_bill'],fill=True)
```

```python
# Pair Plot

sns.pairplot(tips)
```

```python
## HeatMap
corr = tips[['total_bill','tip','size']].corr()
corr  
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
import pandas as pd
sales_df = pd.read_csv('Sales_data.csv')
sales_df.head()
```

```python
## Plot total sales by product
plt.figure(figsize=(10,6))
sns.barplot(x='Product Category', y='Total Revenue', data=sales_df, estimator=sum)
plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
# plt.xticks(rotation=45)
plt.show()
```

```python
## Plot total sales by Region
plt.figure(figsize=(10,6))
sns.barplot(x='Region', y='Total Revenue', data=sales_df, estimator=sum)
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.show()
```
