```python
##Rename columns
df=df.rename(columns={'Country':'state'})
```

```python
## change datatypes
df['Value_new']=df['Salary'].astype(float)
df.head()
```

```python
#another method
df['Salary'] = df['Salary'].replace(58000, 50000)
```

```python
## change datatypes
df['Value_new'] = df['Salary'].fillna(df['Salary'].mean()).astype(int)
df.head()
```

```python
df['New Value']=df['Salary'].apply(lambda x:x*2)
df.head() 
```

```python
## Data Aggregating and grouping
df.head()
```

```python
grouped_mean=df.groupby('Purchased')['Salary'].mean()
print(grouped_mean)
```

```python
grouped_sum=df.groupby(['Purchased','state'])['Salary'].sum()
print(grouped_sum)
```

```python
## aggregate multiple functions
grouped_agg=df.groupby('state')['Salary'].agg(['mean','sum','count'])
grouped_agg
```

```python
## Meging and joining dataframe
# Create sample Dataframes
df1 = pd.DataFrame({'key': ['A','B','C'], 'Value1': [1,2,3]})
df2 = pd.DataFrame({'key': ['A','B','D'], 'Value2': [4,5,6]})
```

```python
df1
```

```python
df2
```

```python
## Merge dataframe on the 'Key' Columns'
pd.merge(df1,df2,on="key",how="inner")
```

```python
pd.merge(df1,df2,on="key",how="outer")
```

```python
pd.merge(df1,df2,on="key",how="left")
```

```python
pd.merge(df1,df2,on="key",how="right")
```
